```python
import nltk
import random
import re

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from textblob import TextBlob

# Download required NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Part 2: Core Dataset

dataset = {
    "hello": ["Hi there!", "Hello!", "Greetings!"],

    "help": ["How can I assist you?",
             "What do you need help with?"],

    "thanks": ["You're welcome!",
               "Glad I could help!"],

    "bye": ["Goodbye!", "See you later!"],

    "how are you": ["I'm doing well, thank you!",
                    "I'm fine! How about you?"],

    "what is your name": ["My name is SimpleBot.",
                          "You can call me SimpleBot."]
}

print("Dataset created successfully!")
```

```python
# Part 3: Core Functions

stop_words = set(stopwords.words('english'))


def preprocess(text):

    # Convert to lowercase and tokenize
    text = text.lower()
    tokens = word_tokenize(text)

    # Remove stopwords and punctuation
    tokens = [token for token in tokens
              if token not in stop_words and token.isalnum()]

    return tokens
```

```python
def handle_ambiguous_input(user_input):

    # Process user input
    user_words = set(preprocess(user_input))

    best_match = None
    best_score = 0

    # Compare with questions in the dataset
    for question in dataset:

        question_words = set(preprocess(question))

        common_words = user_words.intersection(question_words)

        # Calculate matching score
        if len(question_words) > 0:

            score = len(common_words) / len(question_words)

            if score > best_score and score >= 0.5:

                best_score = score
                best_match = question

    # Return a response if a match is found
    if best_match:

        return random.choice(dataset[best_match])

    return "I'm not sure how to respond to that. Could you rephrase it?"
```

```python
def handle_user_input(user_input):

    # Check for empty input
    if not user_input.strip():

        return "Please enter something so I can help you."

    normalized_input = re.sub(r'[^\w\s]', '', user_input.lower())
    normalized_input = " ".join(normalized_input.split())

    # Check for exact matches
    for question in dataset:

        normalized_question = re.sub(
            r'[^\w\s]', '', question.lower()
        )

        normalized_question = " ".join(normalized_question.split())

        if normalized_input == normalized_question:

            return random.choice(dataset[question])

    return handle_ambiguous_input(user_input)
```

```python
def start_chat():

    print("Chatbot: Hello! How can I assist you today?")
    print("Type 'bye', 'exit', or 'quit' to end the chat.")

    while True:

        user_input = input("You: ")

        # Check exit commands
        if user_input.lower().strip() in ["bye", "exit", "quit"]:

            print("Chatbot: Goodbye!")
            break

        # Generate response
        response = handle_user_input(user_input)

        print("Chatbot:", response)
```

```python
# Part 5: Sentiment Analysis Integration

def analyze_sentiment(text):

    try:

        # Analyze the text
        analysis = TextBlob(text)

        # Get sentiment information
        polarity = analysis.sentiment.polarity
        subjectivity = analysis.sentiment.subjectivity

        # Identify sentiment
        if polarity > 0.1:

            sentiment = "POSITIVE"

        elif polarity < -0.1:

            sentiment = "NEGATIVE"

        else:

            sentiment = "NEUTRAL"

        return sentiment, polarity, subjectivity

    except Exception as e:

        print("Error:", e)

        return "UNKNOWN", 0, 0
```

```python
# Function to generate an enhanced response

def enhanced_handle_user_input(user_input):

    # Get the normal chatbot response
    base_response = handle_user_input(user_input)

    # Check for empty input
    if not user_input.strip():

        return base_response

    # Analyze user sentiment
    sentiment, polarity, subjectivity = analyze_sentiment(user_input)

    # Adjust response based on sentiment
    if sentiment == "POSITIVE":

        extra_response = "I'm glad you're feeling positive!"

    elif sentiment == "NEGATIVE":

        extra_response = "I notice you might be feeling down. How can I help?"

    else:

        extra_response = "Thank you for sharing your thoughts!"

    # Combine the responses
    return (
        f"{base_response}\n"
        f"{extra_response}\n"
        f"Sentiment: {sentiment}\n"
        f"Polarity: {polarity:.2f}\n"
        f"Subjectivity: {subjectivity:.2f}"
    )
```

```python
# Testing and Validation

test_sentences = [
    "I love this programming class!",
    "I hate doing homework.",
    "Today is a normal day."
]

# Test sentiment analysis

for sentence in test_sentences:

    sentiment, polarity, subjectivity = analyze_sentiment(sentence)

    print("Sentence:", sentence)
    print("Sentiment:", sentiment)
    print("Polarity:", round(polarity, 2))
    print("Subjectivity:", round(subjectivity, 2))
    print("-" * 40)
```
